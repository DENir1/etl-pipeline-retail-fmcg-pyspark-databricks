```sql
COPY INTO fmcg.gold.fact_orders
FROM (
  SELECT 
    cast(`date` as date) as `date`,
    cast(product_code as string) as product_code,
    cast(customer_code as bigint) as customer_code,       -- Cast to match target type (e.g. INT/STRING)
    cast(sold_quantity as bigint) as sold_quantity        -- Cast to match target type
  FROM '/Volumes/fmcg/gold/parent_incre_data/fact_orders.csv'
)
FILEFORMAT = CSV
FORMAT_OPTIONS (
  'header' = 'true'
);
```
